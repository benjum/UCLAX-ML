Gradient descent is an iterative optimization algorithm for determining the local minimum of a differentiable function.

For a function _J_, the gradient of _J_ with respect to the independent variables gives the direction in which _J_ increases the fastest.  By taking steps opposite to the gradient, one can iteratively move towards lower and lower values until one reaches the minimum of _J_ (ideally).

<img src="https://clairedavid.github.io/ml_in_hep/_images/lec02_1_costSignDirection.png" alt='gradient descent' width='500'/>
[ref: https://clairedavid.github.io/ml_in_hep/_images/lec02_1_costSignDirection.png]

For a $J(\theta)$ that depends on only one variable:
* choose a starting point, $\theta_0$
* take a step towards a new $\theta$:
  * $\theta_{n+1} = \theta_{n} - \eta*dJ/d\theta$
  * here, $\eta$ is a user-chosen step size and $dJ/d\theta$ is the 1D gradient of $J$ (i.e., the derivative)
* continue until the value of $J(\theta)$ stops changing within some low tolerance

```python
# Import our modules
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, interact, interactive_output
from ipywidgets import fixed, VBox, HBox, jslink, Play, FloatSlider, IntSlider

import warnings
warnings.filterwarnings('ignore')
```

We're going to look at a very simple example function:

$$J(\theta) = e^{-\theta}\sin(5\theta)$$

We'll use gradient descent to try to iteratively find a minimum (i.e. a location where $dJ/d\theta=0$).  To do that, we'll need the derivative:

$$\frac{dJ}{d\theta} = -e^{-\theta}\sin(5\theta) + 5e^{-\theta}\cos(5\theta)$$

We'll also need:
* an initial value for $\theta$
* the coefficient $\eta$ that determines how much to vary $\theta$ during a given iteration

```python
def J(t):
    return np.exp(-t) * np.sin(5 * t)
def dJ(t):
    return -np.exp(-t) * np.sin(5 * t) + 5 * np.cos(5 * t) * np.exp(-t)
```

```python
J(0.1), dJ(0.1)
```

```python
def gradient_descent_2(t0, eta=.1, tol=1e-6, num_iters=10):
    ttheory = np.linspace(0.5, 3, 300)
    ytheory = J(ttheory)
    t = [t0]
    i = 0

    fig = plt.figure(figsize=(5,3))
    curve = plt.plot(ttheory, ytheory, 'b-')
    plt.plot(t[-1],J(t[-1]),'ko',markersize=5)
    while i < num_iters:
        t_prev = t[-1]
        grad = dJ(t_prev)
        t_curr = t_prev - eta * grad
        t.append(t_curr)
        if J(t_curr) > 10:
            break
        plt.plot(t_curr,J(t_curr),'ko',markersize=5)
        plt.plot([t_prev,t_curr],[J(t_prev),J(t_curr)],'k--')
        
        if np.abs(t_curr - t_prev) < tol:
            break
        i += 1
    
    fig.show()
```

```python
a = FloatSlider(value=1.3,min=0.5,max=3,description='t0')
b = IntSlider(value=0,min=0,max=50,description='num_iters')
c = FloatSlider(value=0.2,min=0.01,max=1.0,step=0.01,description='eta')
controls = VBox([a,b,c])
plotwidget = interactive_output(gradient_descent_2, {
                       't0':a,
                       'num_iters':b,
                       'eta':c,
                       'tol':fixed(1e-6)});
def achanged(change):
    b.value = 0
a.observe(achanged,'value')

def cchanged(change):
    b.value = 0
c.observe(cchanged,'value')

play = Play(value=0,min=0,max=1000,step=1,interval=1000,description="Press play",disabled=False)
jslink((play, 'value'), (b, 'value'))
gradmov = VBox([controls,plotwidget,play])

display(gradmov)
```
